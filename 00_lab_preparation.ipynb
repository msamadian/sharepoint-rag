# Lab Preparation – SharePoint RAG

This notebook is a short checklist for preparing the lab environment used in the course.

> **Note:** Most steps are performed manually in SharePoint, SQL Server, Windows, or Cursor. The code cells are mainly useful commands that you can copy or run where appropriate.


## 1. Prepare SharePoint Server

We start with an existing SharePoint Server farm.

Create:
- A document library named **Docs**
- A SharePoint list named **Contacts**
- A dedicated **RAG user** with read/write permission to both locations

Later, the ingestion pipeline will populate these locations and read their content.


## 2. Prepare SQL Server

Our embeddings will be stored as vectors in Microsoft SQL Server.

Open SQL Server Management Studio and create an empty database for the project. In this lab we use: **VectorDB**

```python
CREATE DATABASE VectorDB;
GO
```

## 3. Install or Check Python

The ingestion pipeline, query pipeline, and UI are written in Python.

Download Python for Windows from the official Python website if it is not already installed.

During installation, enable **Add Python to PATH**.


```python
python --version
```

## 4. Install or Check Git

Git is used to clone the course project from GitHub.

Install **Git for Windows** if the command below is not recognized.


```python
git --version
```

## 5. Install Visual Studio Code

We use **Visual Studio Code** as the IDE in this course.

Download and install Visual Studio Code for Windows.

The Visual Studio Code **Python extension** is not required to run Python, but it is recommended because it provides features such as code completion, debugging, and Python environment management.

You can also use another Python IDE if you prefer.


## 6. Clone the GitHub Repository

The complete source code is available on GitHub.

Clone the repository instead of creating all project files manually.


```python
git clone https://github.com/msamadian/sharepoint-rag.git
cd sharepoint-rag
```

## 7. Configure `.env`

The project uses a `.env` file to keep environment-specific configuration outside the Python code.

Copy **`.env.example`** to **`.env`**, then configure values such as:

- SharePoint URL
- RAG username and password
- SQL Server and database
- Embedding settings
- Reranker configuration
- LLM endpoint, model, and API key

## 8. Create a Python Virtual Environment

A virtual environment keeps this project's Python packages isolated from other Python projects.


```python
python -m venv .venv
```

### Activate the virtual environment

Run the following command in **Terminal**:


```python
.\.venv\Scripts\Activate.ps1
```

## 9. Install Project Requirements

All required Python packages are listed in `requirements.txt`.

Install them with one command after activating the virtual environment.


```python
pip install -r requirements.txt
```

## 10. Check SQL Server Connectivity Requirement

The Python application connects to SQL Server through ODBC.

Make sure the required **Microsoft ODBC Driver for SQL Server** is installed on the Windows Server before running the ingestion pipeline.
